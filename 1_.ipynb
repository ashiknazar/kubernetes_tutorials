### Kubernetes (K8s)

Kubernetes is a container orchestration platform. It automates the deployment, management, scaling, networking, and recovery of containerized applications.

___

Imagine you have developed an e-commerce application consisting of:
- Frontend – React
- Backend – Flask
- Database – PostgreSQL
- Cache – Redis

You containerize each component using Docker.

```bash
docker run frontend
docker run backend
docker run postgres
docker run redis
```

___

Now imagine the following situations:
1. Your backend container crashes. Who restarts it?
2. Your application receives 10,000 users. Who creates additional backend containers?
3. One server crashes. Who moves your application to another server?
4. How do frontend containers communicate with backend containers?
5. How do you deploy a new application version without significant downtime?

Docker alone does not provide a complete automated solution for all these orchestration requirements.
