### Kubernetes architecture

Main components
- Control plane
- Worker Nodes

___

###  Control plane

The Control Plane manages the entire Kubernetes cluster.

| Component | Responsibility |
|---|---|
| API Server | Entry point for Kubernetes requests |
| Scheduler | Selects a suitable worker node for new Pods |
| Controller Manager | Ensures the cluster maintains its desired state |
| etcd | Stores Kubernetes cluster configuration and state |

### Worker Nodes

Worker nodes are machines where your actual application workloads run.

| Component | Responsibility |
|---|---|
| Kubelet | Communicates with the control plane and manages Pods on its node |
| Container Runtime | Runs containers |
| Kube-proxy | Implements Service networking rules where used |

## Important Kubernetes concepts

___

#### Cluster
A Kubernetes Cluster is a collection of machines working together to run containerized applications.

It consists of:
- One Control Plane.
- One or more Worker Nodes.

___

#### Node
A Node is a physical or virtual machine in a Kubernetes cluster.

___

#### Pod

A Pod is the smallest deployable unit in Kubernetes.

A Pod contains one or more containers.

___

#### Deployment

A Deployment manages the desired number of Pod replicas and handles application updates.

- Suppose you want three Flask backend instances.
  - You create a Deployment with:

```bash
replicas: 3
```

- Kubernetes attempts to maintain three running Pods.
- If one Pod crashes, Kubernetes creates a replacement to restore the desired count.

___

#### Service

- Suppose your Deployment has three Flask Pods.
- Each Pod can have its own IP address. Pod IP addresses can change when Pods are replaced.
- How does your React frontend consistently communicate with the Flask backend?

- This is where a Kubernetes Service comes in.

- A Service provides a stable network endpoint for accessing a group of Pods.

___

#### Namespace

- A Namespace provides logical separation within a Kubernetes cluster.
